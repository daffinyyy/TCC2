# Configurações

```python
DATA = "../dataset_APDDD/dataset_APDDD_completo/data.yaml"
```

```python
YOLO11_NANO = YOLO("runs/detect/11_nano/weights/best.pt") 
YOLO11_SMALL = YOLO("runs/detect/11_small/weights/best.pt") 
YOLO11_MEDIUM = YOLO("runs/detect/11_medium/weights/best.pt") 
YOLO11_LARGE = YOLO("runs/detect/11_large/weights/best.pt") 
YOLO11_EXTRA = YOLO("runs/detect/11_extra/weights/best.pt") 
YOLO12_NANO = YOLO("runs/detect/12_nano/weights/best.pt") 
YOLO12_SMALL = YOLO("runs/detect/12_small/weights/best.pt") 
YOLO12_MEDIUM = YOLO("runs/detect/12_medium/weights/best.pt") 
YOLO12_LARGE = YOLO("runs/detect/12_large/weights/best.pt") 
YOLO12_EXTRA = YOLO("runs/detect/12_extra/weights/best.pt") 
YOLO26_NANO = YOLO("runs/detect/26_nano/weights/best.pt") 
YOLO26_SMALL = YOLO("runs/detect/26_small/weights/best.pt") 
YOLO26_MEDIUM = YOLO("runs/detect/26_medium/weights/best.pt") 
YOLO26_LARGE = YOLO("runs/detect/26_large/weights/best.pt") 
YOLO26_EXTRA = YOLO("runs/detect/26_extra/weights/best.pt") 
RTDETR_LARGE = RTDETR("runs/detect/rtdetr_large/weights/best.pt")
RTDETR_EXTRA = RTDETR("runs/detect/rtdetr_extra/weights/best.pt")
```

```python
models = [YOLO11_NANO, YOLO11_SMALL, YOLO11_MEDIUM, YOLO11_LARGE, YOLO11_EXTRA,
          YOLO12_NANO, YOLO12_SMALL, YOLO12_MEDIUM, YOLO12_LARGE, YOLO12_EXTRA,
          YOLO26_NANO, YOLO26_SMALL, YOLO26_MEDIUM, YOLO26_LARGE, YOLO26_EXTRA, 
          RTDETR_LARGE, RTDETR_EXTRA
         ]
```

# Execução

```python
for model in models:
    print("====================================================")
    print(f"Validando modelo {models.index(model)+1}/{len(models)}...\n")
    validation = model.val()
    print("====================================================\n\n")
print("=========================FIM DE VALIDAÇÃO============================")
```

```python
for model in models:
    results = model.val(
        data=DATA,
        split="val",
        conf=0.50,
        iou=0.50
    )
    
    # F1 de cada classe
    f1_per_class = results.box.f1
    
    # Nomes das classes
    names = results.names
    
    print("=" * 70)
    print("F1 POR CLASSE")
    print("=" * 70)
    
    for i, f1 in enumerate(f1_per_class):
        print(f"{names[i]:25s}: {f1:.4f}")
    
    # F1-macro
    f1_macro = np.mean(f1_per_class)
    
    print("\n" + "=" * 70)
    print(f"F1-MACRO: {f1_macro:.4f}")
    print("=" * 70)
```
