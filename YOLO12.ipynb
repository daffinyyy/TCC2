# Path

```python
data = "dataset_APDDD_completo/data.yaml"
```

# Nano

```python
#hiperparametros e treinamento do modelo
model = YOLO("yolo12n.pt")

results = model.train(
    data=data,
    epochs=300,
    imgsz=640,
    lr0=0.001,
    patience=40,
    optimizer="AdamW",
    device=[0,1],
    name="12_nano"
)
```
